```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
df = pd.merge(movies,credits, on='title', how ='outer')
```

```python
df= df[['id','title','genres','cast', 'crew','keywords','overview']]
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace = True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
import ast
```

```python
def convert(obj):
    l = []
    for i in ast.literal_eval(obj):
       l.append(i['name'])
    return l
```

```python
df['genres'] = df['genres'].apply(convert)
df['keywords'] = df['keywords'].apply(convert)
```

```python
def convert3(obj):
    l = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            l.append(i['name'])
            counter += 1
        else:
            break
    return l
```

```python
df['cast'] = df['cast'].apply(convert3)
```

```python
def fetch_dir(obj):
    l = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l
df['crew'] = df['crew'].apply(fetch_dir)
```

```python
df['overview'] = df['overview'].apply(lambda x:x.split())
```

```python
df['genres'] = df['genres'].apply(lambda x:[i.replace(' ','') for i in x])
df['keywords'] = df['keywords'].apply(lambda x:[i.replace(' ','') for i in x])
df['cast'] = df['cast'].apply(lambda x:[i.replace(' ','') for i in x])
df['crew'] = df['crew'].apply(lambda x:[i.replace(' ','') for i in x])
df.head()
```

```python
df['tags'] = df['genres'] + df['cast'] + df['crew'] + df['overview'] + df['keywords']
df.head()
```

```python
new_df = df[['id','title','tags']]
```

```python
new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
new_df.head()
```

```python
new_df['tags'][1]
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```
